```python
import pandas as pd
import os
import re

AA_STAB = r"C:\Users\user\Downloads\GSE148375_clean\v2_stability_results.csv"
EA_STAB = r"C:\Users\user\Downloads\GSE148812_clean\v2_stability_results.csv"
OUT_DIR = r"C:\Users\user\Desktop\ai causal\causal_project\cross_ancestry\smoking_status"
os.makedirs(OUT_DIR, exist_ok=True)

def strip_suffix(pid):
    return re.sub(r'_\d+$', '', pid)

aa_stab = pd.read_csv(AA_STAB)
ea_stab = pd.read_csv(EA_STAB)

aa_stab["core_name"] = aa_stab["probe_id"].map(strip_suffix)
ea_stab["core_name"] = ea_stab["probe_id"].map(strip_suffix)

print("AA SNPs:", len(aa_stab))
print("EA SNPs:", len(ea_stab))

merged = aa_stab.merge(
    ea_stab, on="core_name", how="inner",
    suffixes=("_AA", "_EA")
)
print("Merged (present in both cohorts):", len(merged))

merged.to_csv(os.path.join(OUT_DIR, "merged_stability_AA_EA_smoking.csv"), index=False)
print("Saved.")
```

```python
import pandas as pd
import os

OUT_DIR = r"C:\Users\user\Desktop\ai causal\causal_project\cross_ancestry\smoking_status"
merged = pd.read_csv(os.path.join(OUT_DIR, "merged_stability_AA_EA_smoking.csv"))

print("=== Joint stability counts ===")
for thresh in [0.5, 0.7, 0.8, 0.9, 1.0]:
    aa_stable = merged["stability_fraction_AA"] >= thresh
    ea_stable = merged["stability_fraction_EA"] >= thresh
    both = (aa_stable & ea_stable).sum()
    aa_only = (aa_stable & ~ea_stable).sum()
    ea_only = (~aa_stable & ea_stable).sum()
    print(f">= {thresh:.0%}  |  AA: {aa_stable.sum()}  EA: {ea_stable.sum()}  "
          f"BOTH: {both}  AA-only: {aa_only}  EA-only: {ea_only}")

corr = merged["stability_fraction_AA"].corr(merged["stability_fraction_EA"])
print(f"\nGenome-wide correlation (stability_fraction_AA vs EA): {corr:.4f}")
```
