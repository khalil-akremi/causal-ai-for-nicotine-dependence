```python
# Extract all three pieces from series matrix in one pass
gsm_ids = []
descriptions = []
statuses = []

with open(series_matrix_path, encoding="utf-8") as f:
    for line in f:
        if "!Sample_geo_accession" in line:
            gsm_ids = [v.strip().strip('"') for v in line.strip().split("\t")[1:]]
        if "!Sample_description" in line:
            descriptions = [v.strip().strip('"') for v in line.strip().split("\t")[1:]]
        if "smoking_status" in line.lower():
            statuses = [v.split(":")[-1].strip().strip('"') for v in line.strip().split("\t")[1:]]

# Build mapping dataframe
meta = pd.DataFrame({
    "gsm_id": gsm_ids,
    "sample_id": descriptions,
    "smoking_status": statuses
})

print(meta.head())
print(meta.shape)
print(meta["smoking_status"].value_counts())
```

```python
# Step 3 — Reorder SNP matrix columns to match meta order
# reorder columns by meta sample order on the transposed matrix (safe reindex)
df_aligned = df.T.reindex(columns=meta["sample_id"].tolist())
df_aligned = df_aligned.T                     # transpose: rows=samples, cols=SNPs
df_aligned.index = meta["sample_id"].tolist() # set index to sample IDs

# Attach smoking status
df_aligned["smoking_status"] = meta["smoking_status"].tolist()

# Verify alignment
print(df_aligned.shape)
print(df_aligned["smoking_status"].value_counts())
print(df_aligned.iloc[:3, :3])
```

```python
# Step 4 — Drop ex-smokers
df_aligned = df_aligned[df_aligned["smoking_status"] != "Ex-smoker"]
print(df_aligned.shape)  # should be (3396, 242902)

# Step 5 — Separate features and target
X_raw = df_aligned.drop(columns=["smoking_status"])
y = (df_aligned["smoking_status"] == "Smoker").astype(int)

print(X_raw.shape)       # should be (3396, 242901)
print(y.value_counts())  # Smoker=1, Non-smoker=0
```

 QC Filter 1 — Missing Rate

```python
# Count NC values per SNP column
missing_rate = (X_raw == "NC").sum() / len(X_raw)

# How many SNPs exceed 10% missing?
print(f"SNPs before filter: {X_raw.shape[1]}")
print(f"SNPs with >10% missing: {(missing_rate > 0.10).sum()}")

# Apply filter
X_qc1 = X_raw.loc[:, missing_rate <= 0.10]
print(f"SNPs after filter: {X_qc1.shape[1]}")
```

QC Filter 2 — Monomorphic SNPs

```python
# Count unique values per SNP column (excluding NC)
def is_monomorphic(col):
    unique_vals = col[col != "NC"].unique()
    return len(unique_vals) <= 1

monomorphic = X_qc1.apply(is_monomorphic)

print(f"SNPs before filter: {X_qc1.shape[1]}")
print(f"Monomorphic SNPs: {monomorphic.sum()}")

# Apply filter
X_qc2 = X_qc1.loc[:, ~monomorphic]
print(f"SNPs after filter: {X_qc2.shape[1]}")
```

QC Filter 3 — Minor Allele Frequency (MAF)

```python
def compute_maf(col):
    # exclude NC
    valid = col[col != "NC"]
    if len(valid) == 0:
        return 0
    # count all alleles
    allele_counts = {}
    for gt in valid:
        for allele in gt:  # each genotype is 2 characters
            allele_counts[allele] = allele_counts.get(allele, 0) + 1
    if len(allele_counts) < 2:
        return 0
    total = sum(allele_counts.values())
    minor_count = min(allele_counts.values())
    return minor_count / total

print("Computing MAF for 202,664 SNPs — this will take a few minutes...")
maf = X_qc2.apply(compute_maf)

print(f"SNPs before filter: {X_qc2.shape[1]}")
print(f"SNPs with MAF < 5%: {(maf < 0.05).sum()}")

X_qc3 = X_qc2.loc[:, maf >= 0.05]
print(f"SNPs after filter: {X_qc3.shape[1]}")
```

```python
# Save QC'd raw data (still strings, before encoding)
try:
    X_qc3.to_parquet("X_qc3.parquet", index=True)
except ImportError:
    X_qc3.to_csv("X_qc3.csv", index=True)
y.to_csv("y.csv")
print("Saved successfully")
```

```python
import numpy as np

def encode_snp_column(col):
    # Replace NC with NaN
    valid = col[col != "NC"]
    
    if len(valid) == 0:
        return pd.Series(np.nan, index=col.index)
    
    # Count all alleles across all patients
    allele_counts = {}
    for gt in valid:
        for allele in gt:
            allele_counts[allele] = allele_counts.get(allele, 0) + 1
    
    if len(allele_counts) < 2:
        return pd.Series(0.0, index=col.index)
    
    # Find minor allele (least frequent)
    minor_allele = min(allele_counts, key=allele_counts.get)
    
    # Count minor allele occurrences per patient
    def count_minor(gt):
        if gt == "NC":
            return np.nan
        return float(gt.count(minor_allele))
    
    return col.map(count_minor)

print("Encoding 31,127 SNPs — this will take several minutes...")
X_encoded = X_qc3.apply(encode_snp_column)

print(f"Shape: {X_encoded.shape}")
print(f"Missing values: {X_encoded.isna().sum().sum()}")
print(f"Unique values sample: {pd.unique(X_encoded.iloc[:, 0].dropna())}")
print(X_encoded.iloc[:3, :3])
```

```python
try:
    X_encoded.to_parquet("X_encoded.parquet")
except ImportError:
    X_encoded.to_csv("X_encoded.csv", index=True)
print("Saved successfully")
print("Saved successfully")
```

```python
from sklearn.impute import KNNImputer

imputer = KNNImputer(n_neighbors=5)

print("Imputing missing values — this will take several minutes...")
X_imputed_array = imputer.fit_transform(X_encoded)

# Convert back to dataframe
X_imputed = pd.DataFrame(
    X_imputed_array,
    index=X_encoded.index,
    columns=X_encoded.columns
)

print(f"Shape: {X_imputed.shape}")
print(f"Missing values remaining: {X_imputed.isna().sum().sum()}")
```

```python
import pyarrow

try:
    X_imputed.to_parquet("X_imputed.parquet")
except (ImportError, ModuleNotFoundError, pyarrow.lib.ArrowKeyError):
    X_imputed.to_csv("X_imputed.csv", index=True)
print("Saved successfully")
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_imputed, 
    y, 
    test_size=0.20,
    random_state=42,
    stratify=y  # ensures same class balance in both splits
)

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train distribution:\n{y_train.value_counts()}")
print(f"y_test distribution:\n{y_test.value_counts()}")
```

```python
from xgboost import XGBClassifier

print("Training XGBoost for feature selection — this will take a few minutes...")

xgb_selector = XGBClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1,          # use all CPU cores
    eval_metric="logloss",
    verbosity=0
)

xgb_selector.fit(X_train, y_train)

# Get feature importances
importances = pd.Series(
    xgb_selector.feature_importances_,
    index=X_train.columns
).sort_values(ascending=False)

print(f"Top 10 most important SNPs:\n{importances.head(10)}")
print(f"\nSNPs with non-zero importance: {(importances > 0).sum()}")
```
