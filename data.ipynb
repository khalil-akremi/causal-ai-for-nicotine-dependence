 QC Filter 1 — Missing Rate

```python
# Count NC values per SNP column
missing_rate = (X_raw == "NC").sum() / len(X_raw)

# How many SNPs exceed 10% missing?
print(f"SNPs before filter: {X_raw.shape[1]}")
print(f"SNPs with >10% missing: {(missing_rate > 0.10).sum()}")

# Apply filter
X_qc1 = X_raw.loc[:, missing_rate <= 0.10]
print(f"SNPs after filter: {X_qc1.shape[1]}")
```

QC Filter 2 — Monomorphic SNPs

```python
# Count unique values per SNP column (excluding NC)
def is_monomorphic(col):
    unique_vals = col[col != "NC"].unique()
    return len(unique_vals) <= 1

monomorphic = X_qc1.apply(is_monomorphic)

print(f"SNPs before filter: {X_qc1.shape[1]}")
print(f"Monomorphic SNPs: {monomorphic.sum()}")

# Apply filter
X_qc2 = X_qc1.loc[:, ~monomorphic]
print(f"SNPs after filter: {X_qc2.shape[1]}")
```

QC Filter 3 — Minor Allele Frequency (MAF)

```python
def compute_maf(col):
    # exclude NC
    valid = col[col != "NC"]
    if len(valid) == 0:
        return 0
    # count all alleles
    allele_counts = {}
    for gt in valid:
        for allele in gt:  # each genotype is 2 characters
            allele_counts[allele] = allele_counts.get(allele, 0) + 1
    if len(allele_counts) < 2:
        return 0
    total = sum(allele_counts.values())
    minor_count = min(allele_counts.values())
    return minor_count / total

print("Computing MAF for 202,664 SNPs — this will take a few minutes...")
maf = X_qc2.apply(compute_maf)

print(f"SNPs before filter: {X_qc2.shape[1]}")
print(f"SNPs with MAF < 5%: {(maf < 0.05).sum()}")

X_qc3 = X_qc2.loc[:, maf >= 0.05]
print(f"SNPs after filter: {X_qc3.shape[1]}")
```

```python
import numpy as np

def encode_snp_column(col):
    # Replace NC with NaN
    valid = col[col != "NC"]
    
    if len(valid) == 0:
        return pd.Series(np.nan, index=col.index)
    
    # Count all alleles across all patients
    allele_counts = {}
    for gt in valid:
        for allele in gt:
            allele_counts[allele] = allele_counts.get(allele, 0) + 1
    
    if len(allele_counts) < 2:
        return pd.Series(0.0, index=col.index)
    
    # Find minor allele (least frequent)
    minor_allele = min(allele_counts, key=allele_counts.get)
    
    # Count minor allele occurrences per patient
    def count_minor(gt):
        if gt == "NC":
            return np.nan
        return float(gt.count(minor_allele))
    
    return col.map(count_minor)

print("Encoding 31,127 SNPs — this will take several minutes...")
X_encoded = X_qc3.apply(encode_snp_column)

print(f"Shape: {X_encoded.shape}")
print(f"Missing values: {X_encoded.isna().sum().sum()}")
print(f"Unique values sample: {pd.unique(X_encoded.iloc[:, 0].dropna())}")
print(X_encoded.iloc[:3, :3])
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_imputed, 
    y, 
    test_size=0.20,
    random_state=42,
    stratify=y  # ensures same class balance in both splits
)

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train distribution:\n{y_train.value_counts()}")
print(f"y_test distribution:\n{y_test.value_counts()}")
```

```python
from xgboost import XGBClassifier

print("Training XGBoost for feature selection — this will take a few minutes...")

xgb_selector = XGBClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1,          # use all CPU cores
    eval_metric="logloss",
    verbosity=0
)

xgb_selector.fit(X_train, y_train)

# Get feature importances
importances = pd.Series(
    xgb_selector.feature_importances_,
    index=X_train.columns
).sort_values(ascending=False)

print(f"Top 10 most important SNPs:\n{importances.head(10)}")
print(f"\nSNPs with non-zero importance: {(importances > 0).sum()}")
```

```python
# Keep top 500 SNPs by importance
top_snps = importances[importances > 0].head(500).index.tolist()

X_train_selected = X_train[top_snps]
X_test_selected = X_test[top_snps]

print(f"X_train_selected shape: {X_train_selected.shape}")
print(f"X_test_selected shape: {X_test_selected.shape}")
```

```python
import optuna
from sklearn.model_selection import StratifiedKFold, cross_val_score
from xgboost import XGBClassifier

optuna.logging.set_verbosity(optuna.logging.WARNING)  # suppress verbose output

def objective_xgb(trial):
    params = {
        "n_estimators": trial.suggest_int("n_estimators", 100, 500),
        "max_depth": trial.suggest_int("max_depth", 3, 8),
        "learning_rate": trial.suggest_float("learning_rate", 0.01, 0.3, log=True),
        "subsample": trial.suggest_float("subsample", 0.6, 1.0),
        "colsample_bytree": trial.suggest_float("colsample_bytree", 0.6, 1.0),
        "min_child_weight": trial.suggest_int("min_child_weight", 1, 10),
        "random_state": 42,
        "n_jobs": -1,
        "eval_metric": "logloss",
        "verbosity": 0
    }
    
    model = XGBClassifier(**params)
    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    scores = cross_val_score(model, X_train_selected, y_train, cv=cv, scoring="roc_auc")
    return scores.mean()

study_xgb = optuna.create_study(direction="maximize")
study_xgb.optimize(objective_xgb, n_trials=50)

print(f"Best XGBoost AUC (CV): {study_xgb.best_value:.4f}")
print(f"Best params: {study_xgb.best_params}")
```

```python
conservative_params = {
    'n_estimators': 437,
    'max_depth': 4,           # shallower trees
    'learning_rate': 0.025408861560855204,
    'subsample': 0.9773162727024731,
    'colsample_bytree': 0.9025120815428063,
    'min_child_weight': 5,    # more conservative
    'reg_alpha': 0.1,         # L1 regularization
    'reg_lambda': 1.0,        # L2 regularization
    'random_state': 42,
    'eval_metric': 'logloss',
    'verbosity': 0,
    'n_jobs': -1
}
```

```python
from sklearn.feature_selection import SelectKBest, chi2
from sklearn.model_selection import StratifiedKFold, cross_val_score
from xgboost import XGBClassifier
from sklearn.metrics import roc_auc_score

# Try stricter feature selection
for k in [100, 200, 500, 1000]:
    selector_k = SelectKBest(chi2, k=k)
    X_train_k = selector_k.fit_transform(X_train, y_train)
    X_test_k  = selector_k.transform(X_test)
    
    model = XGBClassifier(**conservative_params)  # no extra args
    
    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    cv_auc = cross_val_score(model, X_train_k, y_train, cv=cv, scoring="roc_auc").mean()
    
    model.fit(X_train_k, y_train)
    test_auc = roc_auc_score(y_test, model.predict_proba(X_test_k)[:, 1])
    
    print(f"k={k:5d} | CV AUC: {cv_auc:.4f} | Test AUC: {test_auc:.4f} | Gap: {cv_auc - test_auc:.4f}")
```

```python
# Train XGBoost on ALL 31,127 SNPs directly
# Use aggressive subsampling to control overfitting
model_full = XGBClassifier(
    n_estimators=300,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.6,
    colsample_bytree=0.3,  # only 30% of SNPs per tree — key parameter
    min_child_weight=10,
    reg_alpha=0.5,
    reg_lambda=2.0,
    random_state=42,
    eval_metric='logloss',
    verbosity=0,
    n_jobs=-1
)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_auc = cross_val_score(model_full, X_train, y_train, cv=cv, scoring="roc_auc").mean()
print(f"CV AUC (full SNPs): {cv_auc:.4f}")
```
