# Cross-Ancestry Pathway Comparison — Smoking Status

**Purpose:** Compare AA and EA PC-confirmed causal gene sets at the pathway level,
not just the gene level. Individual gene overlap was limited to 1 confirmed shared
gene (DNAH1) out of 33 (AA) and 24 (EA) PC-confirmed genes. This step tests whether
the two ancestry-specific gene sets nonetheless converge on shared biological
pathways — a legitimate, pre-registered comparison (not a result of loosening any
earlier significance threshold).

**Method:** Run Enrichr (GO_Biological_Process_2023, KEGG_2021_Human, Reactome_2022)
separately on AA's and EA's PC-confirmed gene lists. Compare top enriched terms
across both, looking for shared or biologically related pathways even where
member genes differ.

```python
import requests
import pandas as pd

def run_enrichr(genes, description):
    genes_str = '\n'.join(genes)
    payload = {'list': (None, genes_str), 'description': (None, description)}
    r = requests.post('https://maayanlab.cloud/Enrichr/addList', files=payload)
    r.raise_for_status()
    user_list_id = r.json()['userListId']

    libraries = ['GO_Biological_Process_2023', 'KEGG_2021_Human', 'Reactome_2022']
    results = {}
    for lib in libraries:
        q = requests.get(f'https://maayanlab.cloud/Enrichr/enrich?userListId={user_list_id}&backgroundType={lib}')
        q.raise_for_status()
        data = q.json()[lib]
        df = pd.DataFrame(data, columns=['Rank', 'Term', 'P-value', 'Z-score', 'Combined score',
                                          'Genes', 'Adjusted P-value', 'Old P-value', 'Old adjusted P-value'])
        results[lib] = df.sort_values('Adjusted P-value')
    return results

aa_genes_pc = ['ZNF805', 'DVL1', 'LAMA5', 'DNAH5', 'DCHS2', 'USH2A', 'TICAM1', 'CELSR2', 
               'PLEC', 'RYR1', 'FRAS1', 'XIRP1', 'CHD6', 'RNF213', 'ZNF418', 'FREM1',
               'FREM2', 'DNAH1', 'GPR98', 'ASTN1', 'HYDIN', 'SIGLEC1', 'EXO1', 'LZTFL1', 
               'TSC2', 'HMCN1', 'RTN4', 'UTP20', 'CNTRL', 'CCBL2', 'MDN1', 'COL6A6', 'CPAMD8']

ea_genes_pc = ['ABCA8', 'MYBBP1A', 'SMG5', 'SYNE1', 'DICER1', 'PPARGC1B', 'DNAH10', 'DCHS1', 
               'RP11-1055B8.7', 'OBSL1', 'ATP12A', 'KMT2A', 'ZFHX4', 'LRP1B', 'SULT1C2', 
               'FRYL', 'SACS', 'DNAH1', 'PRKDC', 'MAGI1', 'CP', 'OTOF', 'ATP2C2', 'MLLT1']

results_aa = run_enrichr(aa_genes_pc, 'AA_smoking_PC_confirmed')
results_ea = run_enrichr(ea_genes_pc, 'EA_smoking_PC_confirmed')

for lib in ['GO_Biological_Process_2023', 'KEGG_2021_Human', 'Reactome_2022']:
    print(f"\n{'='*70}\n{lib}\n{'='*70}")
    print("\n--- AA top 10 ---")
    print(results_aa[lib].head(10)[['Term', 'Adjusted P-value', 'Genes']].to_string(index=False))
    print("\n--- EA top 10 ---")
    print(results_ea[lib].head(10)[['Term', 'Adjusted P-value', 'Genes']].to_string(index=False))
```
