```python
import pandas as pd
import re

series_matrix_path = r"C:\Users\user\Downloads\GSE148375_series_matrix (1).txt\GSE148375_series_matrix (1).txt"

meta_dict = {}

with open(series_matrix_path, encoding="utf-8") as f:
    for line in f:
        if line.startswith("!Sample_description"):
            meta_dict["sample_id"] = [v.strip('"') for v in line.strip().split("\t")[1:]]
        elif line.startswith("!Sample_characteristics_ch1"):
            values = [v.strip('"') for v in line.strip().split("\t")[1:]]
            # figure out the characteristic name from first value, e.g. "smoking_status: Smoker"
            key = values[0].split(":")[0].strip()
            clean_values = [v.split(":", 1)[1].strip() if ":" in v else v for v in values]
            # avoid collision if same key appears twice
            if key in meta_dict:
                key = key + "_2"
            meta_dict[key] = clean_values

meta_df = pd.DataFrame(meta_dict)
print(meta_df.shape)
print(meta_df.head())
print(meta_df.columns.tolist())
```

```python
import pandas as pd
import os

file_path = r"C:\Users\user\Downloads\GSE148375_processed_data (1).txt\GSE148375_processed_data.txt"
out_dir = r"C:\Users\user\Downloads\GSE148375_clean"

# samples to drop (ex-smokers) -- from cleaned metadata
ex_smoker_ids = set(meta_df.loc[meta_df["smoking_status"] == "Ex-smoker", "sample_id"])
print("Dropping sample IDs:", ex_smoker_ids)

# normalize genotype: sort letters alphabetically, keep NC as is
def normalize_genotype(x):
    if x == "NC":
        return "NC"
    return "".join(sorted(x))

chunksize = 20000
reader = pd.read_csv(file_path, sep="\t", chunksize=chunksize)

out_path = os.path.join(out_dir, "checkpoint0_snp_clean.txt")
first_chunk = True

for chunk in reader:
    id_col = chunk.columns[0]

    # drop indel probes
    chunk = chunk[~chunk[id_col].str.contains("-IND", regex=False)]

    # drop ex-smoker sample columns
    cols_to_drop = [c for c in chunk.columns if c in ex_smoker_ids]
    chunk = chunk.drop(columns=cols_to_drop)

    # normalize genotype codes (vectorized over all geno columns)
    geno_cols = chunk.columns[1:]
    chunk[geno_cols] = chunk[geno_cols].applymap(normalize_genotype)

    chunk.to_csv(out_path, sep="\t", mode="w" if first_chunk else "a",
                 header=first_chunk, index=False)
    first_chunk = False

print("Saved cleaned SNP checkpoint to:", out_path)
```

```python
import pandas as pd
import os

file_path = r"C:\Users\user\Downloads\GSE148375_processed_data (1).txt\GSE148375_processed_data.txt"
out_dir = r"C:\Users\user\Downloads\GSE148375_clean"

indel_codes = {"DI", "ID", "DD", "II"}
ex_smoker_ids = set(meta_df.loc[meta_df["smoking_status"] == "Ex-smoker", "sample_id"])

def normalize_genotype(x):
    if x == "NC":
        return "NC"
    return "".join(sorted(x))

chunksize = 20000
reader = pd.read_csv(file_path, sep="\t", chunksize=chunksize)

out_path = os.path.join(out_dir, "checkpoint0_snp_clean_v2.txt")
first_chunk = True
total_dropped_indel = 0

for chunk in reader:
    id_col = chunk.columns[0]
    geno_cols_all = chunk.columns[1:]

    # drop rows with ANY indel genotype code (more reliable than name matching)
    indel_mask = chunk[geno_cols_all].isin(indel_codes).any(axis=1)
    total_dropped_indel += indel_mask.sum()
    chunk = chunk[~indel_mask]

    # drop ex-smoker sample columns
    cols_to_drop = [c for c in chunk.columns if c in ex_smoker_ids]
    chunk = chunk.drop(columns=cols_to_drop)

    # normalize genotype codes
    geno_cols = chunk.columns[1:]
    chunk[geno_cols] = chunk[geno_cols].map(normalize_genotype)

    chunk.to_csv(out_path, sep="\t", mode="w" if first_chunk else "a",
                 header=first_chunk, index=False)
    first_chunk = False

print("Total indel rows dropped:", total_dropped_indel)
print("Saved corrected checkpoint to:", out_path)
```
